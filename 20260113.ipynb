## 2110번: 공유기 설치

도현이의 집 N개가 수직선 위에 있다. 각각의 집의 좌표는 x1, ..., xN이고, 집 여러개가 같은 좌표를 가지는 일은 없다.

도현이는 언제 어디서나 와이파이를 즐기기 위해서 집에 공유기 C개를 설치하려고 한다. 최대한 많은 곳에서 와이파이를 사용하려고 하기 때문에, 한 집에는 공유기를 하나만 설치할 수 있고, 가장 인접한 두 공유기 사이의 거리를 가능한 크게 하여 설치하려고 한다.

C개의 공유기를 N개의 집에 적당히 설치해서, 가장 인접한 두 공유기 사이의 거리를 최대로 하는 프로그램을 작성하시오.

**정답을 정하는 순간:**

주어진 x를 리스트에 넣고 임의로 C개로 나눈다. 그 중에서 가장 큰 길이가 있으면 갱신. 더 이상 큰 길이가 없으면 break한다.

**gpt가 추천한 사고:**

거리 x를 기준으로 왼쪽부터 설치할 때, C개 이상 설치되면 가능.

```python
N, C = input().split()
N, C = int(N), int(C)
sharing = []

for _ in range(N) :
  sharing.append(int(input()))
```

1 2 4 8 9

0 1 2 3 4

```python
sharing.sort()
left, right = 0, max(sharing)

while True :
  x = (left + right) // 2
  count = 1
  last = sharing[0]

  for i in range(1,N) :
    if sharing[i] - last >= x :
      last = sharing[i]
      count += 1

      if count >= C :
        break
  if count < C :
    right = x-1

  if count >= C :
    left = x+1

  if left > right :
    break

print(right)
```

## 6236번: 용돈 관리

현우는 용돈을 효율적으로 활용하기 위해 계획을 짜기로 하였다. 현우는 앞으로 N일 동안 자신이 사용할 금액을 계산하였고, 돈을 펑펑 쓰지 않기 위해 정확히 M번만 통장에서 돈을 빼서 쓰기로 하였다. 현우는 통장에서 K원을 인출하며, 통장에서 뺀 돈으로 하루를 보낼 수 있으면 그대로 사용하고, 모자라게 되면 남은 금액은 통장에 집어넣고 다시 K원을 인출한다. 다만 현우는 M이라는 숫자를 좋아하기 때문에, 정확히 M번을 맞추기 위해서 남은 금액이 그날 사용할 금액보다 많더라도 남은 금액은 통장에 집어넣고 다시 K원을 인출할 수 있다. 현우는 돈을 아끼기 위해 인출 금액 K를 최소화하기로 하였다. 현우가 필요한 최소 금액 K를 계산하는 프로그램을 작성하시오.

**정답을 만드는 사고:**

N과 M을 입력받는다
하루에 이용할 금액을 받고 money리스트에 넣는다.

제일 먼저 구간을 잡는데
left = max(money)
right = sum(money)로 잡는다
while 문을 돌린다
판단하는 x를 left + right // 2로 두고 시작한다.

count 변수를 두고 기본값은 1로 놓는다.
x원을 기본으로 잡고 for 문을 돌려서 x - money[i]를 한다.
만약 돈이 부족해지면 count를 1 올리고 x원을 더 추가한다.
만약 count가 M을 넘어가면 break
left를 x+1로 바꿔 x값을 올린다

다시 while문을 수행한다. 만약 수행을 하는데 끝까지 했는데도 M보다 count가 작다
그러면 right를 x-1로 바꿔 x값을 내린다

다시 반복문을 수행해서 M과 딱 만나면 break

**gpt가 추천한 사고:**

문제에서 정답은 가능한 K중 최소값
left > right가 되는 순간 마지막으로 가능했던 값이 정답

```python
N, M = input().split()
N, M = int(N), int(M)
money = []

for _ in range(N) :
  money.append(int(input()))
```

```python
left = max(money)
right = sum(money)

while True :
  x = (left + right) // 2
  count = 1
  wallet = x

  for i in range(N) :
    if wallet < money[i] :
      wallet = x
      count += 1
    wallet -= money[i]

    if count > M :
      break

  if count > M :
    left = x+1
  elif count <= M :
    right = x-1

  if left > right :
    break

print(left)
```
