## 5430번: AC

선영이는 주말에 할 일이 없어서 새로운 언어 AC를 만들었다. AC는 정수 배열에 연산을 하기 위해 만든 언어이다. 이 언어에는 두 가지 함수 R(뒤집기)과 D(버리기)가 있다.

함수 R은 배열에 있는 수의 순서를 뒤집는 함수이고, D는 첫 번째 수를 버리는 함수이다. 배열이 비어있는데 D를 사용한 경우에는 에러가 발생한다.

함수는 조합해서 한 번에 사용할 수 있다. 예를 들어, "AB"는 A를 수행한 다음에 바로 이어서 B를 수행하는 함수이다. 예를 들어, "RDD"는 배열을 뒤집은 다음 처음 두 수를 버리는 함수이다.

배열의 초기값과 수행할 함수가 주어졌을 때, 최종 결과를 구하는 프로그램을 작성하시오.

```python
from collections import deque

T = int(input())
result = []

for i in range(T) :
  p = input()
  n = int(input())
  solver = input()[1:-1]

  if solver :
    dq = deque(solver.split(','))
  else :
    dq = deque()

  rev = False
  error = False

  for j in p :
    if j == 'R' :
      rev = not rev
    else :
      if len(dq) == 0 :
        error = True
        break
      if rev == False :
        dq.popleft()
      else :
        dq.pop()

  if error :
    result.append('error')
  else :
    if rev :
      dq.reverse()
    result.append('[' + ','.join(dq) + ']')

for i in result :
  print(i)
```

## 1863번: 스카이라인 쉬운거

도시에서 태양이 질 때에 보이는 건물들의 윤곽을 스카이라인이라고 한다. 스카이라인만을 보고서 도시에 세워진 건물이 몇 채인지 알아 낼 수 있을까? 건물은 모두 직사각형 모양으로 밋밋하게 생겼다고 가정한다.

정확히 건물이 몇 개 있는지 알아내는 것은 대부분의 경우에 불가능하고, 건물이 최소한 몇 채 인지 알아내는 것은 가능해 보인다. 이를 알아내는 프로그램을 작성해 보자.

```python
n = int(input())
stack = []
count = 0

for _ in range(n) :
  x, y = map(int, input().split())

  while stack and y < stack[-1] :
      stack.pop()

  if y > 0 and (not stack or y > stack[-1]) :
    stack.append(y)
    count += 1

print(count)
```
