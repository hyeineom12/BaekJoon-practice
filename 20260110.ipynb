## 11866번: 요세푸스 문제 0

요세푸스 문제는 다음과 같다.

1번부터 N번까지 N명의 사람이 원을 이루면서 앉아있고, 양의 정수 K(≤ N)가 주어진다. 이제 순서대로 K번째 사람을 제거한다. 한 사람이 제거되면 남은 사람들로 이루어진 원을 따라 이 과정을 계속해 나간다. 이 과정은 N명의 사람이 모두 제거될 때까지 계속된다. 원에서 사람들이 제거되는 순서를 (N, K)-요세푸스 순열이라고 한다. 예를 들어 (7, 3)-요세푸스 순열은 <3, 6, 2, 7, 5, 1, 4>이다.

N과 K가 주어지면 (N, K)-요세푸스 순열을 구하는 프로그램을 작성하시오.

**정답이 증가하는 순간은 언제인가?:**

K번째 순서에 해당하면 정답에 추가한 후, 원래 리스트에서 삭제

가장 첫번째 시도에서는 첫번째 값부터 시작하지만 리스트에서 삭제한 후부터는 그 숫자 뒤부터 시작

**gpt에게 보내고 받은 추천 사고:**

앞에서 K-1번 이동 후, 맨 앞이 제거


```python
from collections import deque

N, K = input().split()
N, K = int(N), int(K)

result = []
dq = deque(i+1 for i in range(N))

while len(dq) > 0 :
  for _ in range(K-1) :
    dq.append(dq.popleft())

  result.append(dq.popleft())

print('<'+str(result[0]), end='')
for i in range(1, N) :
  print(',', result[i], end = '')
print('>')
```

## 4949번: 균형잡힌 세상

세계는 균형이 잘 잡혀있어야 한다. 양과 음, 빛과 어둠 그리고 왼쪽 괄호와 오른쪽 괄호처럼 말이다.

정민이의 임무는 어떤 문자열이 주어졌을 때, 괄호들의 균형이 잘 맞춰져 있는지 판단하는 프로그램을 짜는 것이다.

문자열에 포함되는 괄호는 소괄호("()") 와 대괄호("[]")로 2종류이고, 문자열이 균형을 이루는 조건은 아래와 같다.

* 모든 왼쪽 소괄호("(")는 오른쪽 소괄호(")")와만 짝을 이뤄야 한다.
* 모든 왼쪽 대괄호("[")는 오른쪽 대괄호("]")와만 짝을 이뤄야 한다.
* 모든 오른쪽 괄호들은 자신과 짝을 이룰 수 있는 왼쪽 괄호가 존재한다.
* 모든 괄호들의 짝은 1:1 매칭만 가능하다. 즉, 괄호 하나가 둘 이상의 괄호와 짝지어지지 않는다.
* 짝을 이루는 두 괄호가 있을 때, 그 사이에 있는 문자열도 균형이 잡혀야 한다.

정민이를 도와 문자열이 주어졌을 때 균형잡힌 문자열인지 아닌지를 판단해보자.

**언제 오답이 확정되는가?:**

(와 [를 stack에 넣음

), ]를 만나면 pop.

만약 온점이 나왔는데도 len(stack)이 0보다 크면 no를 출력. 0이라면 yes를 출력


**gpt에게 보내고 받은 추천 사고:**

) 또는 ]를 만났는데
* 스택이 비어있음
* 또는 스택 top과 괄호 종류가 다름

-> 즉시 no

입력이 끝났을때 (온점이 나왔을때)

* 스택이 비어있으면 -> yes
* 스택에 뭐라도 남아있으면 -> no



```python
result = []

while True :

  string = input()

  if string == '.' :
    break

  stack = []

  for i in string :
    if i == '(' :
      stack.append('(')
    elif i == '[' :
      stack.append('[')

    if i == ')' :
      if len(stack) == 0 :
        result.append('no')
        break
      if stack[-1] != '(' :
        result.append('no')
        break
      else :
        stack.pop()

    if i == ']' :
      if len(stack) == 0 :
        result.append('no')
        break
      if stack[-1] != '[' :
        result.append('no')
        break
      else :
        stack.pop()

    if i == '.' :
      if len(stack) == 0 :
        result.append('yes')
        break
      else :
        result.append('no')
        break

for i in result :
  print(i)
```
