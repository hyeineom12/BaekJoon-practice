## 2108번: 통계학

수를 처리하는 것은 통계학에서 상당히 중요한 일이다. 통계학에서 N개의 수를 대표하는 기본 통계값에는 다음과 같은 것들이 있다. 단, N은 홀수라고 가정하자.

1. 산술평균 : N개의 수들의 합을 N으로 나눈 값
2. 중앙값 : N개의 수들을 증가하는 순서로 나열했을 경우 그 중앙에 위치하는 값
3. 최빈값 : N개의 수들 중 가장 많이 나타나는 값
4. 범위 : N개의 수들 중 최댓값과 최솟값의 차이


N개의 수가 주어졌을 때, 네 가지 기본 통계값을 구하는 프로그램을 작성하시오.

```python
N = int(input())

nums = []
for _ in range(N) :
  Q = int(input())
  nums.append(Q)
```

```python
mean_nums = round(sum(nums) / len(nums))
print(mean_nums)

#----

nums.sort()
median_nums = nums[N//2]
print(median_nums)

#----

count = {}
for i in range(N) :
  if nums[i] in count :
    count[nums[i]] += 1
  else :
    count[nums[i]] = 1

max_count = max(count.values())
max_count_list = []
for k, v in count.items() :
  if v == max_count :
    max_count_list.append(k)
max_count_list.sort()

if len(max_count_list) >= 2 :
  print(max_count_list[1])
else :
  print(max_count_list[0])

#----

print(max(nums) - min(nums))
```

## 듣보잡

김진영이 듣도 못한 사람의 명단과, 보도 못한 사람의 명단이 주어질 때, 듣도 보도 못한 사람의 명단을 구하는 프로그램을 작성하시오.

```python
N, M = map(int, input().split())

listening = []
for _ in range(N) :
  listening.append(input().rstrip())

seeing = set()
for _ in range(M) :
  seeing.add(input().rstrip())

result = []
for i in listening :
  if i in seeing :
    result.append(i)
result.sort()

print(len(result))
for i in result :
  print(i)
```
