## 2578번: 빙고

오늘은 머리 쉬는 날!

빙고 게임은 다음과 같은 방식으로 이루어진다.

먼저 아래와 같이 25개의 칸으로 이루어진 빙고판에 1부터 25까지 자연수를 한 칸에 하나씩 쓴다

다음은 사회자가 부르는 수를 차례로 지워나간다. 예를 들어 5, 10, 7이 불렸다면 이 세 수를 지운 뒤 빙고판의 모습은 다음과 같다.

차례로 수를 지워가다가 같은 가로줄, 세로줄 또는 대각선 위에 있는 5개의 모든 수가 지워지는 경우 그 줄에 선을 긋는다.

이러한 선이 세 개 이상 그어지는 순간 "빙고"라고 외치는데, 가장 먼저 외치는 사람이 게임의 승자가 된다.

철수는 친구들과 빙고 게임을 하고 있다. 철수가 빙고판에 쓴 수들과 사회자가 부르는 수의 순서가 주어질 때, 사회자가 몇 번째 수를 부른 후 철수가 "빙고"를 외치게 되는지를 출력하는 프로그램을 작성하시오.

**정답 사고:**

binggo 리스트를 만들어 첫번째부터 5번째 줄까지 넣는다. (리스트 안에 리스트가 있는 형태)

result 리스트를 만들어 사회자가 부르는 수를 모두 넣는다.
num = 0

while True를 통해 반복문을 쓴다.

i = result[num]
binggo에 있는 i를 0으로 바꾼다.

만약 모든 행이 0이면 break
또는 모든 열이 0이면 break
또는 대각선이 0이면 break
아니면 num += 1

print(num)

```python
binggo = []
result = []

for _ in range(5) :
  binggo.append(list(map(int, input().split())))

for _ in range(5) :
  result1 = list(map(int,input().split()))
  for i in result1 :
    result.append(i)
```

```python
num = 0
while True :
  x = result[num]
  for i in range(5) :
    for j in range(5) :
      if binggo[i][j] == x :
        binggo[i][j] = 0

  count = 0
  for i in range(5) : # 행 판별
    if all(x == 0 for x in binggo[i]) :
      if count >= 3 :
        break
      count += 1

  for j in range(5) : # 열 판별
    if all(binggo[i][j] == 0 for i in range(5)) :
      if count >= 3 :
        break
      count += 1

  if all(binggo[i][i] == 0 for i in range(5)) :
    count += 1

  if all(binggo[i][4-i] == 0 for i in range(5)) :
    count += 1

  if count >= 3 :
    break
  num += 1

print(num+1)
```

## 14720번: 우유 축제

영학이는 딸기우유, 초코우유, 바나나우유를 좋아한다.

입맛이 매우 까다로운 영학이는 자신만의 우유를 마시는 규칙이 있다.

맨 처음에는 딸기우유를 한 팩 마신다.
딸기우유를 한 팩 마신 후에는 초코우유를 한 팩 마신다.
초코우유를 한 팩 마신 후에는 바나나우유를 한 팩 마신다.
바나나우유를 한 팩 마신 후에는 딸기우유를 한 팩 마신다.
영학이는 우유 축제가 열리고 있는 우유거리에 왔다. 우유 거리에는 우유 가게들이 일렬로 늘어서 있다.

영학이는 우유 거리의 시작부터 끝까지 걸으면서 우유를 사먹고자 한다.

각각의 우유 가게는 딸기, 초코, 바나나 중 한 종류의 우유만을 취급한다.

각각의 우유 가게 앞에서, 영학이는 우유를 사마시거나, 사마시지 않는다.

우유거리에는 사람이 많기 때문에 한 번 지나친 우유 가게에는 다시 갈 수 없다.

영학이가 마실 수 있는 우유의 최대 개수를 구하여라.

```python
N = int(input())
milk = list(map(int, input().split()))
```

```python
younghak = [0, 1, 2] * (N // 3 + 1)

j = 0
count = 0

for i in range(N) :
  if milk[i] == younghak[j] :
    count += 1
    j += 1
    continue
  else :
    continue
print(count)
```
