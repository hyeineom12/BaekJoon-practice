## 1915번: 가장 큰 정사각형

n×m의 0, 1로 된 배열이 있다. 이 배열에서 1로 된 가장 큰 정사각형의 크기를 구하는 프로그램을 작성하시오.

```python
n, m = map(int, input().split())
square = []
for _ in range(n) :
  square.append(list(map(int, input())))
```

```python
max_length = max(map(max, square))

for i in range(1, n) :
  for j in range(1, m) :
    if square[i][j] == 1 :
      square[i][j] = min(
          square[i-1][j],
          square[i][j-1],
          square[i-1][j-1]
      ) + 1

    max_length = max(max_length, square[i][j])

print(max_length**2)
```

해당 문제는 지피티의 도움을 받아 작성하였습니다.
